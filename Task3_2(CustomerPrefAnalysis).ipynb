# **Customer Preference Analysis**

### Analyzing the relationship between the Type of Cuisine and the Restaurant's Rating

```python
import pandas as pd
data = pd.read_csv('Dataset .csv')
```

```python
data.columns
```

##### Calculating the avg rating for each type of cuisine

```python
avg_rating_cuisine = data.groupby('Cuisines')['Aggregate rating'].mean()
#grouping data by 'Cuisines' and then calculating the mean for 'aggregate rating' for each column
print('Average rating for each type of cuisine:')
print(avg_rating_cuisine)
```

### Identifying most popular Cuisines based on votes

##### Grouping the columns based on 'Cuisines' and calculating vote for each

```python
total_votes = data.groupby('Cuisines')['Votes'].sum()
print('Total votes for each cuisine:')
print(total_votes)
```

##### Identifying the most popular cuisines by sorting

```python
most_popular_cuisines = total_votes.sort_values(ascending=False).head(10)
print('Most popular cuisines:')
print(most_popular_cuisines)
```

#### From this, we can understand that cuisines like North Indian - Mughlai, Chinese, Cafe, Fast food etc have received the highest numbers of votes and are the most popular cuisines

### Determining if there are any specific cuisines that tend to receive higher ratings

```python
top_rated_cuisines = avg_rating_cuisine.sort_values(ascending=False)
print('Top rated cuisines:')
print(top_rated_cuisines)
```

#### Thus we can see that cuisines such as Italian, Hawaiian, Seafood, American, Sandwich, Continental Indian, etc have tend to receive higher ratings

#### The identified cuisines with high ratings and most votes reflect specific customer preferences, showcasing popularity.
