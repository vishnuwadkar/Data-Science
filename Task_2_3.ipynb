# **Feature Engineering**

### Extracting additional features from the existing columns

##### Here, we will create new columns based on the length of the restaurant name and address

```python
import pandas as pd
```

```python
data = pd.read_csv('Dataset .csv')
data.columns
```

###### Extracting the length of restaurant name

```python
data['Name Length'] = data['Restaurant Name'].apply(len)
#adding a new columns named 'Name Length' and then calculating the lentgh of each name using len() function
```

###### Extracting the length of restaurant address


```python
data['Address Length'] = data['Address'].apply(len)
#adding a new columns named 'Address Length' and then calculating the length of each Address using len() function
```

```python
print(data[['Restaurant Name', 'Name Length', 'Address', 'Address Length']])
```

### Creating new features by encoding Categorical Variables

##### Here, we will use one hot encoding

###### For 'Has Table booking' column

```python
data['Has Table Booking'] = data['Has Table booking'].apply(lambda x: 1 if x == 'Yes' else 0)
```

###### For 'Has Online delivery' column

```python
data['Has Online Delivery'] = data['Has Online delivery'].apply(lambda x: 1 if x == 'Yes' else 0)
```

```python
print(data[['Restaurant Name', 'Has Table booking','Has Table Booking', 'Has Online delivery', 'Has Online Delivery']].head(10))
```
