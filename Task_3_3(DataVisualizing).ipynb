# **Data Visualization**

```python
import pandas as pd
```

```python
data = pd.read_csv('Dataset .csv')
data.columns
```

###  Creating visualizations to represent the distribution of ratings using different charts

```python
import matplotlib.pyplot as plt
```

##### Creating a histogram to represent the distribution ratings

```python
plt.figure(figsize=(10, 6))
plt.hist(data['Aggregate rating'], bins=10, color='skyblue', edgecolor='black')
plt.title('Distribution of Ratings')
plt.xlabel('Aggregate rating')
plt.ylabel('Frequency')
plt.show()
```

##### Creating a bar plot to represent the distribution of rating

```python
plt.figure(figsize=(10, 6))
plt.bar(data['Aggregate rating'], data['Aggregate rating'], color='skyblue')
plt.title('Distribution of Ratings')
plt.xlabel('Aggregate rating')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

### Comparing the average rating of different cuisines and cities

#### Avg rating of different cuisines

```python
avg_rating_cusines = data.groupby('Cuisines')['Aggregate rating'].mean().sort_values(ascending=False).head(10)
avg_rating_cusines
```

##### Creating a bar graph

```python
plt.figure(figsize=(10, 6))
avg_rating_cusines.plot(kind='bar', color='skyblue')
plt.title('Average Rating by Cuisine')
plt.xlabel('Cuisine')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

#### Avg rating of different cities

```python
avg_rating_cities = data.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False).head(10)
avg_rating_cities
```

```python
plt.figure(figsize=(10, 6))
avg_rating_cities.plot(kind='bar', color='skyblue')
plt.title('Average Rating by Cities')
plt.xlabel('City')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

### Visualizing relationship between various features and target variables

#### Relationship between price range and ratings

```python
plt.figure(figsize=(10, 6))
plt.scatter(data['Price range'], data['Aggregate rating'], color='skyblue', alpha=0.5)
plt.title('Relationship between Price Range and Ratings')
plt.xlabel('Price Range')
plt.ylabel('Average Rating')
plt.grid(True)
plt.show()
```

##### Each point on the scatter plot represents a restaurant, and the plot helps us identify any patterns or correlations between the price range and ratings

##### Relationship between votes and ratings

```python
plt.figure(figsize=(10, 6))
plt.scatter(data['Votes'], data['Aggregate rating'], color='orange', alpha=0.5)
plt.title('Relationship between Votes and Ratings')
plt.xlabel('Votes')
plt.ylabel('Average Rating')
plt.grid(True)
plt.show()
```

##### Each point on the scatter plot represents a restaurant, and the plot helps us identify any patterns or correlations between the number of votes and ratings.
