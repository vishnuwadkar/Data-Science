# Geospatial Analysis

## 1. Visualing the location of restaurants on the map

```python
!pip install gmplot
```

```python
from gmplot import gmplot
import pandas as pd
```

#### Creating a map centered around a specific location

```python
data = pd.read_csv('Dataset .csv')    #loading the dataset
map_center = [data['Latitude'].mean(), data['Longitude'].mean()]
gmap = gmplot.GoogleMapPlotter(map_center[0], map_center[1], 12)
```

#### Adding restaurant locations to map

```python
gmap.scatter(data['Latitude'], data['Longitude'], '#FF0000', size=500, marker=False)
```

#### Saving the map to an HTML file

```python
gmap.draw('restaurant_map.html')
```

## 2. Analyzing the distribution of restaurants across various cities and countries

#### Distribution of restaurants by city

```python
city_distribution = data['City'].value_counts()   #calculating frequency of each unique city on the dataset
print("Distribution of restaurants by city:")
print(city_distribution)
```

#### Distribution of restaurants by Country

```python
city_distribution = data['Country Code'].value_counts()    #calculating frequency of each unique country on the dataset
print("Distribution of restaurants by country:")
print(city_distribution)
```

## 3. Determining correlation between restaurant's location and its rating

#### Scatter plot of rating vs latitude

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))    #sets size of the plot
plt.scatter(data['Latitude'], data['Aggregate rating'], alpha=0.6)    #creates a scatter plot of latitude vs aggregate rating with transparency level of 0.6
plt.xlabel('Latitude')
plt.ylabel('Rating')
plt.title('Latitude vs. Aggregate Rating')
plt.grid(True)
plt.show()   #displaying the plot
```

#### Scatter plot of longitude vs rating

```python
plt.figure(figsize=(10, 6))    #sets size of the plot
plt.scatter(data['Longitude'], data['Aggregate rating'], alpha=0.6)    #creates a scatter plot of latitude vs aggregate rating with transparency level of 0.6
plt.xlabel('Longitude')
plt.ylabel('Rating')
plt.title('Longitude vs. Aggregate Rating')
plt.grid(True)
plt.show()   #displaying the plot
```

#### Calculating correlation between location and rating using correlation coefficient

```python
latitude_correlation = data['Latitude'].corr(data['Aggregate rating'])   #calculating correlation between latitude and rating
longitude_correlation = data['Longitude'].corr(data['Aggregate rating'])   #calculating correlation between longitude and rating
print("Correlation between Latitude and Rating:", latitude_correlation)
print("Correlation between Longitude and Rating:", longitude_correlation)
```


#### Correlation Coefficient
*   0 indicates no correlation
*   1 indicates perfect positive correlation
*   -1 indicates perfect negative correlation



### Thus from the results, we can conclude the following:

#### Restaurant's latitude is almost equal to zero and thus it has no significant impact on the restaurant's rating
#### However, the restaurant's longitude is slightly negative indicating a very weak negative impact on it's rating

## Thus we can conclude that the location of the restaurant has almost no impact on it's rating!!!
