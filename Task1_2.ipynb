```python
data = pd.read_csv("Dataset .csv")
numerical_stats = data.describe()  #calculates the basic stats measures
print(numerical_stats)   #outputing all the stats
```

#### Calculating the median for numerical column

```python
num_cols = data.select_dtypes(include=['number']).columns   #selecting only numerical column
num_median = data[num_cols].median()    #calculatinhg the medan for the selecting column
print(num_median)     #outputing the result
```

## Exploring the distribution for categorial variables

#### Exloring the distribution for variables like "Country Code", "City", and "Cuisines" to help us understand the frequency of different categories in these colmns

```python
cat_col = ["Country Code", "City", "Cuisines"]
for column in cat_col:
    print(f'Distribution of {column}:')
    print(data[column].value_counts())   #calculates the frequency of each unique value
    print("\n")
```

## Identifying top Cuisines and Cities with highest number of restautant

### Top Cuisines

```python
top_cuisines = data["Cuisines"].value_counts().head(10)   #calculates the frequency of each unique cuisine and diplays top 10 cuisines
print(f"Top 10 cuisines with highest number of restaurants are: ")
print(top_cuisines)
```

### Top Cities

```python
top_cities = data["City"].value_counts().head(10)   #calculates the freqeuency of each unique city and displays the top city
print(f"Top 10 Cities with highest number of restaurants are: ")
print(top_cities)
```
