#**Table Booking and Online Delivery**

```python
import pandas as pd
```

### Determining the Percentage of Restaurants that Offer Table Booking and Online Delivery

```python
data = pd.read_csv('Dataset .csv')
```

##### Calculating the restaurants that offer table booking

```python
data.columns
```

```python
table_booking_count = data['Has Table booking'].value_counts()
print("The number of restaurants offering table booking are: ")
print(table_booking_count)
```

###### Calculating percentage

```python
table_booking_perc = data['Has Table booking'].value_counts(normalize=True) * 100  #calculating the percentage through frequency of yes and no
print("The number of restaurants offering table booking are: ")
print(table_booking_perc)
```

#### Therefore, approximately 12 percentage of restaurants offer table booking

### Comparing the average rating of restaurants with and without table booking

###### Let's first filter the restaurant with table booking

```python
with_table_booking = data[data['Has Table booking'] == 'Yes']
```

###### Then we will filter the restaurant without table booking

```python
without_table_booking = data[data['Has Table booking'] == 'No']
```

##### Calculating the average for each group

```python
avg_with_table_booking = with_table_booking['Aggregate rating'].mean()
avg_without_table_booking = without_table_booking['Aggregate rating'].mean()
```

```python
print(f"Average rating for restaurants with table booking: {avg_with_table_booking}")
print(f"Average rating for restaurants without table booking: {avg_without_table_booking}")
```

#### Thus from the results, we can conclude that restaurants that have facility of table booking have higher ratings than the restaurants without table booking.

### Analyzing availibility of online delivery among restaurants with different prices

```python
online_delivery = pd.crosstab(data['Has Online delivery'], data['Price range'], normalize = 'index')*100
# here we did the cross tabulation of 'has online delivery ' column and 'price range' column
print("Availbility of online delivery among restaurants with different prices: ")
print(online_delivery)
```

#### from this we can conclude the following

1.   For price range 1: 28.6% of the restaurants offers online delivery
2.   For price range 2: 52.47% of the restaurants offers online delivery
3.   For price range 3: 16.77% of the restaurants offers online delivery
4.   For price range 4: 2.16% of the restaurants offers online delivery



### Thus we can understand that availability of online delivery varies according to price ranges. Price Range 2 has the highest percentage of restaurants offering online delivery, while price eange 1 and price range 4 have the lowest percentages.
