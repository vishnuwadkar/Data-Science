```python
data = pd.read_csv("Dataset .csv")   #this is a function for loading the dataset from a csv file
```

<h3>Identifying the number of rows and columns</h3>

```python
nums_rows, nums_cols = data.shape    #the shape function here returns the number of rows and columns in the dataset
print(f"The dataset contains: {nums_rows} rows and {nums_cols} columns.")
```

<h3>Checking for missing values</h3>

```python
missing_values = data.isnull().sum()   #calculating the total number of missing values in each column
print("Missing value in each column is: ")
print(missing_values)
```

<h5>Thus there are 9 missing values in Cuisines column</h5>

<h5>These missing values need to be handled</h5>

<h3>Handling missing values</h3>

<h5>For handling the missing values, we can either fill the missing values with mean or we can drop the entire row with missing value</h5>

```python
data.dropna(inplace=True)  #handling missing values only for numerics values

missing_values_after = data.isnull().sum()   #calculating the total number of missing values in each column
print("Missing value in each column is: ")
print(missing_values_after)
```

```python
data["Country Code"] = data["Country Code"].astype(int)  #converts specific column to int
```

```python
data["Average Cost for two"] = data["Average Cost for two"].astype(int)  #converts specific column to int
```

```python
data["Price range"] = data["Price range"].astype(int)  #converts specific column to int
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

<h5>Analyzing the distribution</h5>

```python
plt.figure(figsize=(10, 6))   #sets size of the plot
sns.histplot(data["Aggregate rating"], bins=10, kde=True)  #creates histogram of target variable with KDE(Kernel Density Estimate) overlay
plt.title("Distribution of Aggregate Rating")  #setting title of the plot
#setting the labels
plt.xlabel("Aggregate Rating")
plt.ylabel("Frequency")

#displaying the plot
plt.show()
```

<h3>Identifying the class imbalances</h3>

```python
class_count = data["Aggregate rating"].value_counts()  #calculates the frequency of each class in target variable
print(f"Class distribution of aggregate rating: ")
print(class_count)
```

##### we can understand from this data that majority of the restaurants has 0 ratings that is they have not received any ratings.
##### After that, 3.0 - 3.9 ratings are quite common. This is the average ratings of the restaurants
##### Restaurants with 4 or higher ratings are less common and indicates highly rated restaurants.
##### Restaurants with rating less than 3 are very few and indicates poorly rayed restaurants
