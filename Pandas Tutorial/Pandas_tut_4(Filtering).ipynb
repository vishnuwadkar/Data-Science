#Filtering

```python
import pandas as pd
```

```python
bios = pd.read_csv('https://raw.githubusercontent.com/KeithGalli/complete-pandas-tutorial/master/data/bios.csv')
```

```python
bios
```

```python
bios.info()
```

#### To filter all the athletes above 220cm hieght



```python
bios.loc[bios['height_cm']>220]    #providing a condition to loc function
```

```python
bios.loc[bios['height_cm']>220, ['name', 'height_cm']]    #outputing only the required columns
```

#### This is filtering based on a condition

### A simpler way to do this

```python
bios[bios['height_cm']>220][['name','height_cm']]
```

#### To apply multiple conditions, we use parenthesis and and, or operators

```python
bios[(bios['height_cm']>215) & (bios['born_country']=='USA')][['name','height_cm','born_country']]
#this gives the data of atheletes whose height is greater than 215cm and born country is USA
```

### String methods

```python
bios[bios['name'].str.contains("Krishna")]      #to filter by strings
#here we can provide additional attribute 'case' which can be true or false to change it case sensitive or not
#another attr that can be used is na = False, which removes the na values
```

```python
bios[bios['name'].str.contains("Krishna | Rama")]   #regex syntax used for OR
```

##### Find athletes born in a city staring with a vowel

```python
# using r before the condition is a regular expression (regex syntax)
vowel_cities = bios[bios['born_city'].str.contains(r"^[aeiouAEIOU].*", na=False)]
vowel_cities
# to turn regex off, we can pass an attr regex=False, which will turn off the regex
```

##### Find atheletes whose name contains exactly two vowels

```python
two_vowels = bios[bios['name'].str.contains(r"[aeiouAEIOU]{2}", na= False)]
two_vowels
```

##### Find athletes born in year after 2000


```python
born_19xx = bios[bios['born_date'].str.contains(r"^20", na=False)]
born_19xx
```

##### Find athletes whose city contains exactly 5 letters

```python
city_5letters = bios[bios['born_city'].str.contains(r"^.{5}$", na=False)]
city_5letters
```

### To pass a list as .contains, we use .isin() method

```python
bios[bios['born_country'].isin(['IND', 'RUS'])]
```

##### To find athletes born in PUNE

```python
bios[bios['born_city'].str.contains("Pune", na=False)]
```

To find athletes born in Maharashtra

```python
bios[bios['born_region'].str.contains("Maharashtra", na=False)]
```

```python
two_vowels = bios[bios['born_country'].str.contains(r"USA", na= False) & bios['name'].str.startswith(r"Keith", na= False)]
two_vowels
# gives names starting with Keith and born in USA
```
