# Adding Deleting Columns

```python
import pandas as pd
```

```python
coffee = pd.read_csv('https://raw.githubusercontent.com/KeithGalli/complete-pandas-tutorial/master/warmup-data/coffee.csv')
coffee
```

#### Lets add a new column 'price'

```python
coffee['price'] = 10        # adds a new column 'price' with a deafult value 10
coffee
```

#### To spcifically set values, we use numpy's conditional method

```python
import numpy as np
coffee['price'] = np.where(coffee['Coffee Type'] == 'Espresso', 15, 20)
# syntax is: where(case,if true,else)
coffee
```

##### To drop specific index, we use

```python
coffee.drop(0)
```

#### For dropping a columns, we use

```python
coffee.drop(columns=['price'])
```

##### However this does not actually modifies it

##### To actually modify the data

```python
#coffee.drop(columns=['price'],inplace=True)
#coffee
```

#### To add a new column with some math calculation

```python
coffee['revenue'] = coffee['Units Sold']*coffee['price']
coffee
```

### To rename a column

```python
coffee.rename(columns={'price':'unit_price'},inplace=True)
# old value : new value
coffee
```

```python
bios = pd.read_csv('https://raw.githubusercontent.com/KeithGalli/complete-pandas-tutorial/master/data/bios.csv')
```

```python
bios_new = bios.copy()      #copies to new variable
bios_new
```

#### To access a part of the entire name from the full name

```python
bios_new['first name'] = bios_new['name'].str.split(' ').str.get(0)
bios_new
```

#### To work with dates, convert them to date-time format

```python
bios_new['born_date'] = pd.to_datetime(bios_new['born_date'])
bios_new['died_date'] = pd.to_datetime(bios_new['died_date'])
bios_new
# thus type of data is changed to datetime
```

```python
bios_new.dtypes
```

#### After converting to datatime type, we can use the datetime functions using .dt similar to .str

```python
bios_new['born_year'] = bios_new['born_date'].dt.year   #adding a new column named born year
bios_new
```

```python
bios_new[['name','born_year']].head()
```

### To save the file

```python
bios_new.to_csv('bios_new.csv', index = "False")    #index arg is imp for so it doesn't save the values from the ouput
```

```python
bios_new['height_category'] = bios_new['height_cm'].apply(lambda x: 'short' if x < 160 else ('Average' if x<180 else 'tall'))
bios_new.head(10)
```

### Writing custom functions

```python
def athlete_category(row):
    if row['height_cm'] < 175 and row['weight_kg'] < 70:
        return 'Light Weight'
    elif row['height_cm'] < 190 and row['weight_kg'] <= 90:
        return 'Middle Weight'
    else:
        return 'Heavy weight'
```

```python
bios_new['athlete_category'] = bios_new.apply(athlete_category, axis = 1)       #axis 1 for rows, 0 for columns
bios_new
```
