# Introduction to Dataframes

### importing pandas

```python
import pandas as pd
```

#### **Dataframe** is the main data structure of Pandas library
#### They are simply tables with wide variety of functions
#### Allows to work with spreadsheets

```python
df = pd.DataFrame([[1,2,3],[4,5,2],[9,8,3]], columns = ["A","B","C"])  # a simple dataframe with 2D array
df.head()
```

##### To acces only the first row

```python
df.head(1)
```

##### To acces bottom two rows

```python
df.tail(2)
```

##### By default, index are numerical. But we can specify the index while crating the dataframe

```python
df = pd.DataFrame([[1,2,3],[4,5,2],[9,8,3]], columns = ["X","Y","Z"], index = ["p","q","r"])
```

```python
df
```

```python
df.columns #displays all the columns
```

```python
df.index    #displays all the indexes
```

```python
df.info()     #to get all information about the dataframe
```

```python
df.describe()     #gives meaningful (statistical information) information about the data
```

```python
df.nunique()   #gives number of the unique values in each column
```

```python
df["X"].unique()    #gives unique values from a specific column
```

```python
df.shape      #gives the number of rows and column
```

```python
df.size    #gives the size of the dataframe
```
