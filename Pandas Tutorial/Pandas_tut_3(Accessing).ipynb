#### To acces specific values, we use loc and iloc functions

##### loc -> allows us to filter by rows and columns

```python
df.loc[rows, columns]
df.loc[0]   #gives the first row of the dataframe
df.loc[[0,1,2]]   #gives the first three rows
df.loc[0:2]   #gives the first three rows
df.loc[4:12]   #gives the row 4 to 12
df.loc[0:3,["col1","col3","col5"]]  #accesing rows and columns together
df.loc[0:3,"col1":"col3"]  #accesing rows and columns together
```

##### iloc -> similar to loc but uses index location

```python
df.iloc[0:5,[1,2]]   #onlu index can be passed
df.index = df["Day"]    #making the values of column 'Day' as indexes so that we can use them directly as indexes
df.iloc[0:5,"Monday":"Thursday"]   #accesing rows and columns together
```

### Therefore loc allows us to access rows and columnd whereas iloc also allows us to get rows and columns but only using index values

## To modify any value in the dataset

```python
df.loc[3,"col2"] = value    #changing the value of col2 at 3rd index to 'value'
# we can also modify mutliple values using slicing method
```

### Better way to this is using .at[] and .in

```python
df.at[3,"col2"] = value    #changing the value of col2 at 3rd index to 'value'
```

#### However, this only works for a single value

### To acces a specific column

```python
df["col 1"]  #if the column has a space
df.col1     #if the column name is single word
```

### To sort data, we use .sort_values function

```python
df.sort_values("col_name")    #sorts in ascending order
df.sort_values("col_name", ascending=False)    #sorts in descending order
# we can also use multiple parameters to sort
df.sort_values(["col1","col2"])    #sorts col1 in ascending order and col2 in ascending order if there are similar values in col1
df.sort_values(["col1","col2"], ascending=[True,False])    #sorts col1 in ascending order and col2 in descending order
```

### For iterating among rows and columns

```python
for rows, cols in df.iterrows():
    print(rows)
    print(cols["col1"])
```

##### This gives one row at a time however, the performance is compromised
