# **Price Range Analyisis**

### Determining the most common price common price range among all restaurants

```python
import pandas as pd
```

```python
data = pd.read_csv('Dataset .csv')
data.columns
```

```python
common_price_range = data['Price range'].value_counts().idxmax()
# here, .value_counts() function calculates the frequency of each range and then .idxmax() returns the range with highest frequency
print(f"Most common price range is {common_price_range}")
```

### Calculating the average rating for each price range

```python
avg_rating = data.groupby('Price range')['Aggregate rating'].mean()
print("Average rating for each price range is: ")
print(avg_rating)
```

### Identifying the color that represents th highest Average rating among different price ranges

###### The price with the highest average average rating

```python
highest_avg_pr = avg_rating.idxmax()
print(f"The price range with the highest average rating is {highest_avg_pr}")
```

###### Identifying color respresenting highest avg rating

```python
color_highest_rating = data[data['Price range'] == highest_avg_pr]['Rating color'].mode().iloc[0]
print(f"The color representing the highest average rating is {color_highest_rating}")
```

#### Thus we found out the price range with highest rating and the color representing the highest rating
